# **Array Functions**

```python
a1=np.random.random((3,3))*100
a1=np.round(a1)
a1
```

```python
np.max(a1,axis=0) #axis-0 for 'col' wise
np.max(a1,axis=1) #axis-1 for 'row' wise
np.sum(a1)#min,max,sum,prod
```

```python
print(np.mean(a1)) #mean,median,std,var
print(np.median(a1,axis=0))
print(np.std(a1))
print(np.var(a1))
```

```python
#dot product
np.dot(a1,a2)
```

```python
#log and epxonents
np.log(a1)
np.exp(a1)
```

```python
#round,floor,ceil
np.round(np.random.random((2,3))*100)
```

```python
#trigonometric funs
np.sin(a1)
```
